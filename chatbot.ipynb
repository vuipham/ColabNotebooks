Retriver Tool

```python
from pydantic import BaseModel, Field
from typing import Type, Optional
from langchain.tools import BaseTool
from langchain_core.callbacks import CallbackManagerForToolRun

```

```python
class RetriveInput(BaseModel):
  query: str = Field("", description = "Search query string entered by the user")
```

```python
class RetrieveTool(BaseTool):
    name: str = "retrieve_tool"
    description: str = (
        "Tool to answer questions about labor regulations, privacy policies, and holiday lists"
    )
    args_schema: Type[BaseModel] = RetriveInput
    return_direct: bool = True

    _vector_store: Type[Chroma]
    _k: int

    def __init__(self, vectorstores, **kwargs) -> None:
        super().__init__(**kwargs)
        self._vector_store = vectorstores
        self._k = kwargs.get("k", 4)

    def _run(
        self,
        query: str,
        run_manager: Optional[CallbackManagerForToolRun] = None,
    ) -> str:
        results = self._vector_store.similarity_search(query, k=self._k)
        documents = [doc.page_content for doc in results]

        return "\n\n".join(documents)
```

```python
tool = RetrieveTool(vectorstores=vectorstores)

print(tool.invoke({"query": "Nội quy lao động công ty "}))
```

```python
# def chat(user_message, history):
#   if history is None:
#     history = []

#   messages = history.copy()

#   messages.append({"role": "user", "content": user_message})

#   result = agent.invoke({"messages": messages})

#   bot_reply = result['messages'][-1].content

#   #  add to history
#   history.append({"role": "user", "content": user_message})
#   history.append({"role": "assistant", "content": bot_reply})

#   return "", history

def safe_text(x):
    """Chuyển mọi thứ về string để tránh Gradio Error."""
    try:
        if isinstance(x, str):
            return x
        return str(x)
    except:
        return "⛔ Không thể hiển thị nội dung trả về."


def chat(user_message, history):
    if history is None:
        history = []

    try:
        # copy history cũ
        messages = history.copy()
        messages.append({"role": "user", "content": safe_text(user_message)})

        # gọi agent
        result = agent.invoke({"messages": messages})

        # lấy trả lời cuối
        bot_reply = result["messages"][-1].content
        bot_reply = safe_text(bot_reply)

        # update history
        history.append({"role": "user", "content": safe_text(user_message)})
        history.append({"role": "assistant", "content": bot_reply})

        return "", history

    except Exception as e:
        err_msg = f"⚠️ Lỗi: {safe_text(e)}\nBạn có thể hỏi lại câu khác."
        history.append({"role": "assistant", "content": err_msg})
        return "", history
```

```python
import gradio as gr
```

```python
with gr.Blocks() as demo:
    gr.Markdown("Company X's internal virtual assistant")
    chatbot = gr.Chatbot(type="messages", height=600  )
    with gr.Row():
        txt = gr.Textbox(
            show_label=False,
            placeholder="Nhập tin nhắn..."
        )
    txt.submit(chat, [txt, chatbot], [txt, chatbot])

demo.launch(share=True, debug=True)
```
